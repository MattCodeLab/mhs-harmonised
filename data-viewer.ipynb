```python
import duckdb
from pathlib import Path
def run_query(sql:str):
    con = duckdb.connect()
    df = con.execute(sql).df()
    return df

BASE_DIR = Path("outputs/stage2_harmonised")
OUT_BASE_DIR = Path("outputs/further-processing")
```

```python
# Private Sector Credit
file = "banking_system/credit_and_financing/credit_to_the_private_sector.parquet"
out_file = "credit_to_the_private_sector.parquet"

dir = BASE_DIR / file
out_dir = OUT_BASE_DIR / out_file


df = run_query(f"""
            
            SELECT
                * EXCLUDE ('frequency')
            FROM '{dir}'
            WHERE frequency = 'Monthly'
            ORDER BY date
               """)

df.to_parquet(out_dir, compression='snappy')
```

```python
# MFRS 9
file = "banking_system/asset_quality_and_provisions/mfrs9_financing.parquet"
out_file = "mfrs9_financing.parquet"

dir = BASE_DIR / file
out_dir = OUT_BASE_DIR / out_file


df = run_query(f"""
            
            SELECT
                * EXCLUDE (bank, frequency)
            FROM '{dir}'
            WHERE bank = 'Jumlah / Total'
            AND frequency = 'Monthly'
            ORDER BY date
               """)

df.to_parquet(out_dir, compression='snappy')
```

```python
# MFRS 9
file = "banking_system/asset_quality_and_provisions/mfrs9_financing.parquet"
out_file = "mfrs9_financing.parquet"

dir = BASE_DIR / file
out_dir = OUT_BASE_DIR / out_file


df = run_query(f"""
            
            SELECT
                * EXCLUDE (bank, frequency)
            FROM '{dir}'
            WHERE bank = 'Jumlah / Total'
            AND frequency = 'Monthly'
            ORDER BY date
               """)
df.to_parquet(out_dir, compression='snappy')
```

```python
# MFRS 9 — Islamic
file = "banking_system/asset_quality_and_provisions/mfrs9_financing_islamic.parquet"
out_file = "mfrs9_financing_islamic.parquet"

dir = BASE_DIR / file
out_dir = OUT_BASE_DIR / out_file


df = run_query(f"""
            
            SELECT
                * EXCLUDE (industry, bank, frequency)
            FROM '{dir}'
            WHERE industry = 'Jumlah / Total'
            AND frequency = 'Monthly'
            ORDER BY date
               """)

df.to_parquet(out_dir, compression='snappy')
df.melt(id_vars="date", var_name="indicator", value_name="value")
```
